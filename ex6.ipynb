```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('Bird.png')
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis('off')
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis('off')
plt.show()
```

```python
sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

sobelx = cv2.convertScaleAbs(sobelx)
sobely = cv2.convertScaleAbs(sobely)

plt.figure(figsize=(12, 6))

plt.subplot(121)
plt.imshow(sobelx, cmap='gray')
plt.title('Sobel-X Edge Map')
plt.axis('off')

plt.subplot(122)
plt.imshow(sobely, cmap='gray')
plt.title('Sobel-Y Edge Map')
plt.axis('off')

plt.show()
```

```python
sobel_combined = cv2.magnitude(sobelx.astype(np.float32),
                               sobely.astype(np.float32))

sobel_combined = cv2.convertScaleAbs(sobel_combined)

# Display combined result
plt.imshow(sobel_combined, cmap='gray')
plt.title('Sobel Combined Edge Map')
plt.axis('off')
plt.show()
```

```python
prewitt_x = np.array([
    [-1, 0, 1],
    [-1, 0, 1],
    [-1, 0, 1]
], dtype=np.float32)

prewitt_y = np.array([
    [-1, -1, -1],
    [ 0,  0,  0],
    [ 1,  1,  1]
], dtype=np.float32)

prewittx = cv2.filter2D(gray, cv2.CV_64F, prewitt_x)
prewitty = cv2.filter2D(gray, cv2.CV_64F, prewitt_y)

prewittx = cv2.convertScaleAbs(prewittx)
prewitty = cv2.convertScaleAbs(prewitty)

plt.figure(figsize=(12, 10))

plt.subplot(121)
plt.imshow(prewittx, cmap='gray')
plt.title('Prewitt-X Edge Map')
plt.axis('off')

plt.subplot(122)
plt.imshow(prewitty, cmap='gray')
plt.title('Prewitt-Y Edge Map')
plt.axis('off')

plt.show()
```

```python
prewitt_combined = cv2.magnitude(prewittx.astype(np.float32),
                                prewitty.astype(np.float32))

prewitt_combined = cv2.convertScaleAbs(prewitt_combined)
plt.figure(figsize=(12, 10))
plt.subplot(313)
plt.imshow(prewitt_combined, cmap='gray')
plt.title('Prewitt Combined Edge Map')
plt.axis('off')
plt.show()
```

```python
laplacian = cv2.Laplacian(gray, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)

plt.figure(figsize=(10, 5))

plt.subplot(121)
plt.imshow(gray, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(122)
plt.imshow(laplacian, cmap='gray')
plt.title('Laplacian Edge Detection')
plt.axis('off')

plt.show()
```

```python
canny = cv2.Canny(gray, 100, 200)

plt.figure(figsize=(10, 5))

plt.subplot(121)
plt.imshow(gray, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(122)
plt.imshow(canny, cmap='gray')
plt.title('Canny Edge Detection')
plt.axis('off')

plt.show()
```

```python
roberts_x = np.array([
    [1, 0],
    [0, -1]
], dtype=np.float32)

roberts_y = np.array([
    [0, 1],
    [-1, 0]
], dtype=np.float32)

robertsx = cv2.filter2D(gray, cv2.CV_32F, roberts_x)
robertsy = cv2.filter2D(gray, cv2.CV_32F, roberts_y)

roberts_combined = cv2.magnitude(robertsx, robertsy)

robertsx = cv2.convertScaleAbs(robertsx)
robertsy = cv2.convertScaleAbs(robertsy)
roberts_combined = cv2.convertScaleAbs(roberts_combined)

plt.figure(figsize=(12, 8))

plt.subplot(221)
plt.imshow(gray, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(222)
plt.imshow(robertsx, cmap='gray')
plt.title('Roberts X')
plt.axis('off')

plt.subplot(223)
plt.imshow(robertsy, cmap='gray')
plt.title('Roberts Y')
plt.axis('off')

plt.subplot(224)
plt.imshow(roberts_combined, cmap='gray')
plt.title('Roberts Combined')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
#Comparision of All filters
plt.figure(figsize=(16, 8))

# 1. Original
plt.subplot(2, 4, 1)
plt.imshow(img[:,:,::-1])
plt.title("Original Image")
plt.axis("off")

# 2. Grayscale
plt.subplot(2, 4, 2)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale")
plt.axis("off")

# 3.Sobel
plt.subplot(2, 4, 3)
plt.imshow(sobel_combined, cmap='gray')
plt.title("Sobel Edge")
plt.axis("off")

# 4. Prewitt
plt.subplot(2, 4, 4)
plt.imshow(prewitt_combined, cmap='gray')
plt.title("Prewitt Edge")
plt.axis("off")

# 5. Laplacian
plt.subplot(2, 4, 5)
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian Edge")
plt.axis("off")

# 6. Canny
plt.subplot(2, 4, 6)
plt.imshow(canny, cmap='gray')
plt.title("Canny Edge")
plt.axis("off")

# 7. Roberts
plt.subplot(2, 4, 7)
plt.imshow(roberts_combined, cmap='gray')
plt.title("Roberts Edge")
plt.axis("off")

plt.tight_layout()
plt.show()
```
